```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Income Data.csv")
df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
cols = ["income", "education", "prestige"]

summary = pd.DataFrame({
    "Minimum": df[cols].min(),
    "1st Quartile": df[cols].quantile(0.25),
    "Median": df[cols].median(),
    "Mean": df[cols].mean(),
    "3rd Quartile": df[cols].quantile(0.75),
    "Maximum": df[cols].max(),
}).T

summary.round(2)
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(df["prestige"], bins=10, edgecolor="black")
axes[0].set_title("Histogram of Prestige")
axes[0].set_xlabel("Prestige")
axes[0].set_ylabel("Frequency")

axes[1].hist(df["education"], bins=10, edgecolor="black")
axes[1].set_title("Histogram of Education")
axes[1].set_xlabel("Education")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].scatter(df["income"], df["education"])
axes[0].set_title("Income vs Education")
axes[0].set_xlabel("Income")
axes[0].set_ylabel("Education")

axes[1].scatter(df["education"], df["prestige"])
axes[1].set_title("Education vs Prestige")
axes[1].set_xlabel("Education")
axes[1].set_ylabel("Prestige")

plt.tight_layout()
plt.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
income_var = np.var(df["income"], ddof=1)   # sample variance
income_std = np.std(df["income"], ddof=1)   # sample standard deviation

print(f"Variance of income: {income_var:.2f}")
print(f"Standard deviation of income: {income_std:.2f}")
```

9. Explain what variance measures

11. As a data analyst, if you calculate the income variance and it is 0, what does that mean? Is it good or bad?

**9. What variance measures:** Variance measures how spread out the values are around the mean. It is the average of the squared distances between each value and the mean. For *income*, the variance is about 597.07, so the standard deviation (its square root, in the original units) is about 24.44.

**10. What a high variance suggests:** A high variance means the values are widely spread out: many data points are far from the mean, so the data is less consistent and the mean alone does not represent individual values well. Here, an income standard deviation of 24.44 against a mean of 41.87 (a range from 7 to 81) shows that incomes in this dataset vary a lot between jobs.

**11. If income variance is 0:** A variance of 0 means every value is identical: every person in the dataset has exactly the same income. It is not "good" or "bad" in itself, but as an analyst it is a red flag. In real income data it almost certainly means a data problem (for example, a column filled with a default value or a data-entry/import error). Even if it were true, a variable with no variation carries no information. It can't explain differences between people and can't be correlated with anything, so it is useless for analysis.
