2. Create a dataframe from the data contained in *income_data.csv* using `pandas`

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("income_data.csv")
df.head()
```

3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *education*

```python
r, p = stats.pearsonr(df["income"], df["education"])
print(f"Income vs Education: r = {r:.4f}, p-value = {p:.4e}")
```

3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *prestige*

```python
r, p = stats.pearsonr(df["income"], df["prestige"])
print(f"Income vs Prestige: r = {r:.4f}, p-value = {p:.4e}")
```

4. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *education* and *prestige*

```python
r, p = stats.pearsonr(df["education"], df["prestige"])
print(f"Education vs Prestige: r = {r:.4f}, p-value = {p:.4e}")
```

6. Pearson's correlation coefficient OR covariance can both be used to measure the direction of the relationship between two variables (e.g., education and income). Explain why Pearson's correlation coefficient may be better than covariance.

Covariance only tells us the **direction** of a relationship (positive or negative). Its size depends on the units and scale of the variables. For example, income in dollars gives a much larger covariance than income in thousands of dollars, even though the relationship is identical. So a covariance of 500 could be strong or weak, and covariances between different pairs of variables cannot be compared.

Pearson's correlation coefficient is covariance **standardized** by the standard deviations of both variables. That makes it unit-free and always between −1 and +1, so it tells us both the **direction and the strength** of the linear relationship. It can be compared across different pairs of variables (e.g., we can say education–prestige, r = 0.85, is stronger than income–education, r = 0.72), and it comes with a p-value to test significance. That makes it more interpretable and more useful than covariance.
